JOINING:
a) Concat - Quickest way (row/column-wise) - Stacking , Duplication

b) Merge - database style joins (sql joiner) ~ Powerful.

c) Join - shorthand for merge (index)

## MERGE

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'Stud_ID':[1,2,3,4],
    'Name':['Alice','Bob','Jane','Moses'],
    'Subject':['Math','Eng','Hist','Sci']
})
```

```python
df2 = pd.DataFrame({
    'Stud_ID':[3,4,5,6],
    'Score':['90','85','88','92'],
    'Grade':['A','B','B+','A-']
})
```

```python
df1
```

```python
df2
```

DEFAULT WAY

```python
pd.merge(left=df1,right=df2)
```

TYPES OF JOIN/MERGE:

```python
pd.merge(df1,df2,on='Stud_ID',how='inner')
```

```python
df1.merge(df2,on='Stud_ID',how='inner')
```

```python
pd.merge(df1,df2,on='Stud_ID',how='left')
```

```python
pd.merge(df1,df2,on='Stud_ID',how='right')
```

```python
pd.merge(df1,df2,on='Stud_ID',how='outer')
```

LEFT ON & RIGHT ON

```python
df1 = pd.DataFrame({
    'Stud_ID':[1,2,3,4],
    'Name':['Alice','Bob','Jane','Moses'],
    'Subject':['Math','Eng','Hist','Sci']
})

df2 = pd.DataFrame({
    'ID':[3,4,5,6],
    'Score':['90','85','88','92'],
    'Grade':['A','B','B+','A-']
})
```

```python
pd.merge(df1,df2,left_on='Stud_ID',right_on='ID',how='inner')
```

LEFT INDEX & RIGHT INDEX

```python
pd.merge(df1,df2,left_index=True,right_index=True)
```

## JOIN

```python
df1 = pd.DataFrame({
    'Stud_ID':[1,2,3,4],
    'Name':['Alice','Bob','Jane','Moses'],
    'Subject':['Math','Eng','Hist','Sci']
})

df2 = pd.DataFrame({
    'ID':[3,4,5,6],
    'Score':['90','85','88','92'],
    'Grade':['A','B','B+','A-']
})
```

```python
df1.set_index('Stud_ID',inplace=True)
```

```python
df1
```

```python
df2.set_index('ID',inplace=True)
```

```python
df2
```

```python
df1.join(df2,how='inner')           # Alternative to LEFT ON & RIGHT ON (USE SET_INDEX FUNCTION)
```

### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user). 

### Step 3. Assign it to a variable called users.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user"
users = pd.read_csv(url, sep="|", names=["user_id", "age", "gender", "occupation", "zip_code"], index_col="user_id")

users.head()
```

```python
users.info()
```

### Step 4. Discover what is the mean age per occupation

```python
# Convert age column to numeric
users["age"] = pd.to_numeric(users["age"])
```

```python
users.groupby("occupation")["age"].mean().sort_values(ascending=False)
```

### Step 5. Discover the Male ratio per occupation and sort it from the most to the least

```python
a = users[users["gender"] == "M"].groupby("occupation").size() 
```

```python
b = users.groupby("occupation").size()
```

```python
a / b
```

### Step 6. For each occupation, calculate the minimum and maximum ages

```python
users.groupby("occupation")["age"].agg(["min", "max"])
```

### Step 7. For each combination of occupation and gender, calculate the mean age

```python
users.groupby(["occupation", "gender"])["age"].mean()
```

### Step 8.  For each occupation present the percentage of women and men

```python
male = users[users["gender"] == "M"].groupby("occupation").size()
female = users[users["gender"] == "F"].groupby("occupation").size() 
total = users.groupby("occupation").size()

(male / total) * 100
(female / total) * 100
```

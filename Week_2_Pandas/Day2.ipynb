### Data Selection In Series Objects

```python
import pandas as pd
```

```python
# Review of Series Objects:
num = [1,2,3,4,5]
num_series = pd.Series(num)
num_series
```

```python
# Indexing:
num_series[]
```

Series Object ~ Dictionary: 

```python
# Py Expressions/Methods/Syntax

# 1. 'in' (Check if a key/index is in a series object):

cities = {'Nakuru':130000, 'Voi':25000, 'Kwale':12500}
pop = pd.Series(cities)

'Voi' in pop
```

```python
# 2. Retrieve all keys: ~ keys() method
pop.keys()
```

```python
# 3. Retrieve a list of key-value pairs: ~ items() method
list(pop.items())
```

```python
# 4.Add a key-value pair: ~ new index
pop['Nairobi'] = 300000
pop
```

Series Object ~ 1D Array: 

```python
# Numpy Array Methods ~ 'Indexing Format'

# 1. Slicing (Using 'Keys') ~ Inclusive
pop['Nakuru':'Nairobi']
```

```python
# 2. Slicing (Using 'Int/Values') ~ Exclusive
pop[0:3]
```

```python
# 3. Masking (~ Range)
pop[(pop > 125000) & (pop < 350000)]
```

```python
# 4. Specific Indexing
pop[['Nakuru','Nairobi']]
```

Indexers (loc and iloc)

```python
# Confusion between Slicing & Indexing (Implicit & Explicit)

data = pd.Series(['a', 'b', 'c'], index=[1, 3, 5])
data
```

```python
# Indexing ~ Explicit
data[1]
```

```python
# Slicing ~ Implicit (Default Indexes)
data[1:3]
```

```python
# Pandas Solution: "SPECIAL INDEXERS" ~ Attributes.

# 1. loc attribute (Changes 'Default Index' to 'Specific')
data.loc[1]
```

```python
data.loc[1:3]
```

```python
# 2. iloc attribute (Changes 'Specific' to 'Default Index')
data.iloc[1]
```

```python
data.iloc[1:3]
```

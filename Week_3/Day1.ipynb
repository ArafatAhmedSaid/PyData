```python
data['date'] = pd.to_datetime(data['date'])
```

```python
# Create a figure
plt.figure(figsize=(12,4))

# Line Plot
plt.plot(data['date'],data['ALUMINUM'],linestyle='--',color='orange',label='Aluminum')
plt.plot(data['date'],data['COPPER'],color='black',label='Copper')
plt.title('Aluminum & Copper Stock Prices', fontsize=12,fontweight='bold',loc='left',pad=15)
plt.xlabel('DATE')
plt.ylabel('Stock Prices')
plt.legend()
plt.show()
```

```python
# Add Subplot
fig,ax = plt.subplots(nrows=1,ncols=2,figsize=(12,4))
ax[0].plot(data['date'],data['ALUMINUM'],linestyle='--',color='orange',label='Aluminum')
ax[1].plot(data['date'],data['COPPER'],color='black',label='Copper')
ax[0].set_title('Al Prices')
ax[1].set_title('Cu Prices')
ax[0].set_ylabel('Al Prices')
ax[0].legend()
plt.show()
```
